```python
import numpy as np
from obspy import Stream, Trace

# Function to create a Stream object from seismic data
def create_segy_stream(seismic_data, sample_interval_ms):
    stream = Stream()
    
    # Ensure number of traces is within valid range
    num_traces = min(seismic_data.shape[1], 32768)
    
    for i, data in enumerate(seismic_data.T[:num_traces]):  # Transpose to get each trace as a column
        trace = Trace()
        trace.data = data
        trace.stats.network = "NETWORK"
        trace.stats.station = f"STATION_{i + 1}"
        trace.stats.location = "00"
        trace.stats.channel = "CH"
        trace.stats.delta = sample_interval_ms / 1000.0
        trace.stats.segy = {'sample_interval_in_milliseconds': sample_interval_ms}
        stream.append(trace)

    return stream

# Normalize the grayscale data to have values between -1 and 1
normalized_seismic_data = 2 * (seismic_traces - np.min(seismic_traces)) / np.ptp(seismic_traces) - 1

# Sample interval in milliseconds
sample_interval_ms = 2.0

# Output file paths for two halves
#output_segy_path_1 = r'C:\Users\lmag0269\OneDrive - The University of Sydney (Staff)\Coral_Sea_seismic\FromMaria\SEISMIC_DATA\I00041377\Section_3_Interpreted_Seismic_Line 74_1130_TVF_half1.sgy'
#output_segy_path_2 = r'C:\Users\lmag0269\OneDrive - The University of Sydney (Staff)\Coral_Sea_seismic\FromMaria\SEISMIC_DATA\I00041377\Section_3_Interpreted_Seismic_Line 74_1130_TVF_half2.sgy'

#output_segy_path_1 = r'C:\Users\lmag0269\OneDrive - The University of Sydney (Staff)\Coral_Sea_seismic\FromMaria\SEISMIC_DATA\I00041377\Section_4_Interpreted_Seismic_Line_1128_NTG_half1.sgy'
#output_segy_path_2 = r'C:\Users\lmag0269\OneDrive - The University of Sydney (Staff)\Coral_Sea_seismic\FromMaria\SEISMIC_DATA\I00041377\Section_4_Interpreted_Seismic_Line_1128_NTG_half2.sgy'

#output_segy_path_1 = r'C:\Users\lmag0269\OneDrive - The University of Sydney (Staff)\Coral_Sea_seismic\FromMaria\SEISMIC_DATA\I00041377\Section_6_Interpreted_Seismic_Line_1121_NTG_half1.sgy'
#output_segy_path_2 = r'C:\Users\lmag0269\OneDrive - The University of Sydney (Staff)\Coral_Sea_seismic\FromMaria\SEISMIC_DATA\I00041377\Section_6_Interpreted_Seismic_Line_1121_NTG_half2.sgy'

output_segy_path_1 = r'C:\Users\lmag0269\OneDrive - The University of Sydney (Staff)\Coral_Sea_seismic\FromMaria\SEISMIC_DATA\I00041377\Section_7_Interpreted_Seismic_Line_1115_NTG_TVD_DGF_half1.sgy'
output_segy_path_2 = r'C:\Users\lmag0269\OneDrive - The University of Sydney (Staff)\Coral_Sea_seismic\FromMaria\SEISMIC_DATA\I00041377\Section_7_Interpreted_Seismic_Line_1115_NTG_TVD_DGF_half2.sgy'

# Split the normalized seismic data into two halves
half_index = normalized_seismic_data.shape[1] // 2
seismic_data_half1 = normalized_seismic_data[:, :half_index]
seismic_data_half2 = normalized_seismic_data[:, half_index:]

# Create and write SEGY files for each half
stream_half1 = create_segy_stream(seismic_data_half1, sample_interval_ms)
stream_half1.write(output_segy_path_1, format='SEGY', data_encoding=1)  # Assuming data_encoding 1 for IBM floating point
print(f"SEGY file saved to {output_segy_path_1}")

stream_half2 = create_segy_stream(seismic_data_half2, sample_interval_ms)
stream_half2.write(output_segy_path_2, format='SEGY', data_encoding=1)  # Assuming data_encoding 1 for IBM floating point
print(f"SEGY file saved to {output_segy_path_2}")

```
