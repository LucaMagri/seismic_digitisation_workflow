```python
import numpy as np

# Define the input and output file names

input_file_name = r'C:\Users\lmag0269\OneDrive - The University of Sydney (Staff)\Coral_Sea_seismic_profiles_navigation\Hill_1992\A-B_Hill92.txt'
output_file_name = r'C:\Users\lmag0269\OneDrive - The University of Sydney (Staff)\Coral_Sea_seismic_profiles_navigation\Hill_1992\A-B_Hill92.nav'


# Read the data from the input file
values_col2 = []
values_col3 = []
with open(input_file_name, 'r') as file:
    for line in file:
        columns = line.strip().split(',')
        if len(columns) >= 2:
            try:
                value_col2 = float(columns[0])
                value_col3 = float(columns[1])
                values_col2.append(value_col2)
                values_col3.append(value_col3)
            except ValueError:
                continue

# Define the starting and ending values for the increment --------- CHANGE THIS FOR EACH NAV (CHECK PROFILES ON KINGDOM ALREADY LAODED)
start_value = 1.0
end_value = 1555.0

# Generate the incrementing values for column 1
incrementing_values = np.arange(start_value, end_value + 1, 1.0)

# Interpolate the values for columns 2 and 3
num_increments = len(incrementing_values)
if len(values_col2) > 1:
    values_col2_interp = np.linspace(values_col2[0], values_col2[1], num_increments)
    values_col3_interp = np.linspace(values_col3[0], values_col3[1], num_increments)
else:
    values_col2_interp = [0] * num_increments  # Fallback if not enough data
    values_col3_interp = [0] * num_increments  # Fallback if not enough data

# Prepare the content for the new .nav file with even spacing
nav_lines = []
for i in range(num_increments):
    nav_lines.append(f"{incrementing_values[i]:<10.1f} {values_col2_interp[i]:<15.6f} {values_col3_interp[i]:<15.6f}\n")

# Write the content to the output file
with open(output_file_name, 'w') as file:
    file.writelines(nav_lines)

print(f"Data has been written to {output_file_name}")
```
